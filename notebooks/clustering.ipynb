```python
import pandas as pd
from sklearn.cluster import KMeans
import joblib

# load your dataset (use dataset1 ideally)
df = pd.read_csv("Food_Delivery_Times.csv")

# features for clustering
X = df[["Distance_km", "Delivery_Time_min"]]
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[["Distance_km", "Delivery_Time_min"]])

# train
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_scaled)
```

```python
df['cluster'] = kmeans.labels_
print(df.groupby("cluster")[["Distance_km", "Delivery_Time_min"]].mean())
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Distance_km"], df["Delivery_Time_min"], c=df["cluster"])

plt.xlabel("Distance (km)")
plt.ylabel("Delivery Time (min)")
plt.title("Clusters")

plt.show()
```

```python
joblib.dump(kmeans, "kmeans.pkl")
```
